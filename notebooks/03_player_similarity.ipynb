# 03 — Oyuncu Benzerliği ve Transfer Scout

```python
import sys; sys.path.insert(0, '..')
from dotenv import load_dotenv; load_dotenv('../.env')

from src import db
from src.analysis import transfer
from src.viz.radars import player_radar, comparison_radar
import matplotlib.pyplot as plt

LEAGUE, SEASON = 'Süper Lig', '2024-2025'
player_stats = db.get_player_stats(league=LEAGUE, season=SEASON)
print(f'{len(player_stats)} oyuncu yüklendi')
```

```python
# Hedef oyuncu için benzerlik
target = player_stats.dropna(subset=['xg']).sort_values('xg', ascending=False).iloc[0]['player']
print('Hedef:', target)
similar = transfer.find_similar_players(player_stats, target, n=8, min_minutes=450)
similar
```

```python
# Radar
fig = player_radar(player_stats, target)
plt.show()
```

```python
# Karşılaştırma — hedef + en benzer 2 oyuncu
players_to_compare = [target] + similar.head(2)['player'].tolist()
fig = comparison_radar(player_stats, players_to_compare)
plt.show()
```

```python
# Değer/Performans Sıralaması
vp = transfer.value_performance_ratio(player_stats)
cols = ['player', 'team', 'position', 'minutes', 'xg_xa_per90', 'market_value', 'value_score']
cols = [c for c in cols if c in vp.columns]
vp[cols].head(15)
```
