```python
import sys; sys.path.insert(0, '..')
from dotenv import load_dotenv; load_dotenv('../.env')

import pandas as pd
import matplotlib.pyplot as plt
from src import db

print('Sezonlar:', db.get_available_seasons())
print('Ligler:', db.get_available_leagues())
```

```python
LEAGUE = 'Süper Lig'
SEASON = '2024-2025'

team_stats = db.get_team_stats(league=LEAGUE, season=SEASON)
print('Takım sayısı:', len(team_stats))
team_stats.head()
```

```python
# xG vs Gerçek Gol
if 'xg' in team_stats.columns and 'goals_for' in team_stats.columns:
    df = team_stats.dropna(subset=['xg', 'goals_for'])
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.scatter(df['xg'], df['goals_for'], s=120, alpha=0.7)
    for _, row in df.iterrows():
        ax.annotate(row['team_name'], (row['xg'], row['goals_for']), fontsize=8)
    lim = [min(df['xg'].min(), df['goals_for'].min()), max(df['xg'].max(), df['goals_for'].max())]
    ax.plot(lim, lim, 'r--', label='y=x')
    ax.set_xlabel('xG'); ax.set_ylabel('Gerçek Gol'); ax.legend(); ax.set_title('xG vs Gerçek Gol')
    plt.show()
```

```python
player_stats = db.get_player_stats(league=LEAGUE, season=SEASON)
print(f'{len(player_stats)} oyuncu')
player_stats.head()
```
