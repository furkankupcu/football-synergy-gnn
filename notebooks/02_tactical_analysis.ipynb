# 02 — Taktiksel Analiz
PPDA proxy, pressing stilleri, atak verimliliği.

```python
import sys; sys.path.insert(0, '..')
from dotenv import load_dotenv; load_dotenv('../.env')

from src import db
from src.analysis import tactical
from src.viz.style_plots import ppda_bar

LEAGUE, SEASON = 'Süper Lig', '2024-2025'
team_stats = db.get_team_stats(league=LEAGUE, season=SEASON)
report = tactical.build_tactical_report(team_stats)
report.sort_values('ppda')
```

```python
# Pressing stil dağılımı
if 'pressing_style' in report.columns:
    counts = report['pressing_style'].value_counts()
    counts.plot(kind='pie', autopct='%1.1f%%', figsize=(8,8))
```

```python
# PPDA bar
if report['ppda'].notna().any():
    fig = ppda_bar(report, team_col='team_name')
    fig.show()
```

```python
# Atak verimliliği — kim xG'sinden fazla/az gol atıyor?
eff = tactical.attack_efficiency(team_stats)
eff[['team_name', 'goals_for', 'xg', 'goal_xg_diff']].dropna().sort_values('goal_xg_diff', ascending=False)
```
